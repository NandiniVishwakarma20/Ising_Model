```python
import numpy as np
import matplotlib.pyplot as plt

# Load data
data = np.loadtxt('ising_observables.txt')
T = data[:, 0]
E = data[:, 1]
M = data[:, 2]
C = data[:, 3]
chi = data[:, 4]

# Create a figure and subplots
plt.figure(figsize=(10, 8))

# Average Energy vs Temperature
plt.subplot(2, 2, 1)
plt.plot(T, E, label='Average Energy')
plt.xlabel('Temperature (T)')
plt.ylabel('<E>')
plt.title('Average Energy vs Temperature')
# plt.grid(True)
plt.legend()

# Average Magnetization vs Temperature
plt.subplot(2, 2, 2)
plt.plot(T, M, label='Average Magnetization')
plt.xlabel('Temperature (T)')
plt.ylabel('<M>')
plt.title('Average Magnetization vs Temperature')
# plt.grid(True)
plt.legend()

# Specific Heat vs Temperature
plt.subplot(2, 2, 3)
plt.plot(T, C, label='Specific Heat')
plt.xlabel('Temperature (T)')
plt.ylabel('C')
plt.title('Specific Heat vs Temperature')
# plt.grid(True)
plt.legend()

# Susceptibility vs Temperature
plt.subplot(2, 2, 4)
plt.plot(T, chi, label='Susceptibility')
plt.xlabel('Temperature (T)')
plt.ylabel('Chi')
plt.title('Susceptibility vs Temperature')
# plt.grid(True)
plt.legend()

# Adjust layout and show
plt.tight_layout()
plt.show()
```
