```python
import matplotlib.pyplot as plt
import numpy as np

# Load the data
data = np.loadtxt('ising_free_bc_observables.txt')

# Extract columns
T = data[:, 0]      # Temperature
E = data[:, 1]      # Energy
M = data[:, 2]      # Magnetization
C = data[:, 3]      # Specific Heat
chi = data[:, 4]    # Magnetic Susceptibility

# Create a figure with subplots
plt.figure(figsize=(14, 10))

# Plot 1: Temperature vs Energy
plt.subplot(2, 2, 1)
plt.plot(T, E, color='red')
plt.xlabel('Temperature (T)', fontsize=12)
plt.ylabel('Energy per Spin <E>', fontsize=12)
plt.title('Energy vs Temperature', fontsize=14)
# plt.grid(True, alpha=0.3)

# Plot 2: Temperature vs Magnetization
plt.subplot(2, 2, 2)
plt.plot(T, M, color='orange')
plt.xlabel('Temperature (T)', fontsize=12)
plt.ylabel('Magnetization <|M|>', fontsize=12)
plt.title('Magnetization vs Temperature', fontsize=14)
# plt.grid(True, alpha=0.3)

# Plot 3: Temperature vs Specific Heat
plt.subplot(2, 2, 3)
plt.plot(T, C, color='blue')
plt.xlabel('Temperature (T)', fontsize=12)
plt.ylabel('Specific Heat (C)', fontsize=12)
plt.title('Specific Heat vs Temperature', fontsize=14)
# plt.grid(True, alpha=0.3)

# Plot 4: Temperature vs Magnetic Susceptibility
plt.subplot(2, 2, 4)
plt.plot(T, chi, color='green')
plt.xlabel('Temperature (T)', fontsize=12)
plt.ylabel('Magnetic Susceptibility (χ)', fontsize=12)
plt.title('Magnetic Susceptibility vs Temperature', fontsize=14)
# plt.grid(True, alpha=0.3)

# Add a title for the entire figure
plt.suptitle('2D Ising Model Results (L=' + str(20) + ')', fontsize=16)

# Adjust layout
plt.tight_layout(rect=[0, 0, 1, 0.95])

# Save the figure
plt.savefig('ising_model_results.png', dpi=300, bbox_inches='tight')

# Show the plot
plt.show()

print("Visualization complete! Results saved as 'ising_model_results.png'")
```
